```python
#!/usr/bin/env python3
"""
Build a single concatenated line per session using the already-copied exports.

Assumes transcripts are in:
  ~/.Noxi/noxi_exports/transcripts/
with filenames like:
  001_expert.audio.transcript.annotation.csv
  001_novice.audio.transcript.annotation.csv

Outputs:
  ~/.Noxi/noxi_exports/combined_sentence/<SESSION>_sentence.txt
Each file contains one line:
  <SPK1> text, <SPK2> text, <SPK1> text, ...

Usage:
  python combine_from_exports.py
  # or specify locations:
  python combine_from_exports.py --transcripts ~/.Noxi/noxi_exports/transcripts --out ~/.Noxi/noxi_exports/combined_sentence
"""
import argparse
import csv
import re
from pathlib import Path

def parse_session_from_filename(name: str) -> str:
    # Expect prefix like "001_expert.audio.transcript.annotation.csv"
    m = re.match(r"^([0-9]{3,})_(expert|novice)\.audio\.transcript\.annotation\.csv$", name)
    if not m:
        return ""
    return m.group(1)

def read_transcript_csv(path: Path, tag: str, encoding="utf-8"):
    rows = []
    with path.open("r", encoding=encoding, newline="") as f:
        reader = csv.reader(f, delimiter=";")
        for row in reader:
            if len(row) < 3:
                continue
            try:
                start = float(row[0].strip().replace(",", "."))
                end = float(row[1].strip().replace(",", "."))
            except ValueError:
                # skip header or malformed
                continue
            text = row[2].strip()
            if text:
                rows.append((start, end, tag, text))
    return rows

def build_sentence(rows, include_times=False):
    parts = []
    for start, end, tag, text in rows:
        if include_times:
            parts.append(f"{tag} [{start:.2f};{end:.2f}] {text}")
        else:
            parts.append(f"{tag} {text}")
    return " ".join(parts)
```

```python

ap = argparse.ArgumentParser()
ap.add_argument("--transcripts", default="~/.Noxi/noxi_exports/transcripts", help="Path to exported transcripts folder.")
ap.add_argument("--out", default="~/.Noxi/noxi_exports/combined_sentence", help="Output folder for one-line conversations.")
encoding = "utf-8"
include_times = False

transcripts_dir = Path("./transcripts").expanduser().resolve()
out_dir = Path("./combined_sentence").expanduser().resolve()
out_dir.mkdir(parents=True, exist_ok=True)

# Group files by session
by_session = {}
for path in transcripts_dir.glob("*_*.audio.transcript.annotation.csv"):
    session = parse_session_from_filename(path.name)
    if not session:
        continue
    role = "expert" if "expert.audio.transcript.annotation.csv" in path.name else "novice"
    by_session.setdefault(session, {})[role] = path

if not by_session:
    print(f"No transcript files matched under {transcripts_dir}")

total = 0
for session, files in sorted(by_session.items()):
    expert = files.get("expert")
    novice = files.get("novice")
    if not expert or not novice:
        # require both to build a dialogue
        continue
    rows = []
    rows += read_transcript_csv(expert, "<SPK1>", encoding)
    rows += read_transcript_csv(novice, "<SPK2>", encoding)
    rows.sort(key=lambda x: (x[0], x[1]))

    sentence = build_sentence(rows, include_times=include_times)
    out_path = out_dir / f"{session}_sentence.txt"
    out_path.write_text(sentence, encoding="utf-8")
    print(f"Wrote: {out_path} ({len(rows)} utterances)")
    total += 1

print(f"Done. Sessions processed: {total}. Output dir: {out_dir}")
```

```python
#!/usr/bin/env python3
"""
Collect all combined one-line conversation texts into a single CSV.

Assumes you have per-session files like:
  ~/.Noxi/noxi_exports/combined_sentence/001_sentence.txt
  ~/.Noxi/noxi_exports/combined_sentence/026_sentence.txt
  ...

The CSV will have two columns:
  file_name,text
where file_name is the session folder name (e.g., "001") and text is the
entire concatenated conversation line.

Usage:
  python export_combined_to_csv.py
  # or specify paths:
  python export_combined_to_csv.py --in ~/.Noxi/noxi_exports/combined_sentence \
                                   --out ~/.Noxi/noxi_exports/combined_all.csv
"""
import argparse
import csv
import re
from pathlib import Path

def session_from_filename(name: str) -> str:
    # Accept leading digits as session id (e.g., 001, 026, 150)
    m = re.match(r"^([0-9]{3,})", name)
    return m.group(1) if m else name
indir = Path("./combined_sentence").expanduser().resolve()
outfile = Path("./conversations.csv").expanduser().resolve()
outfile.parent.mkdir(parents=True, exist_ok=True)

rows = []
for p in sorted(indir.glob("*_sentence.txt")):
    session = session_from_filename(p.stem)  # stem like "001_sentence"
    # remove trailing suffix if present
    session = session.replace("_sentence", "")
    text = p.read_text(encoding=encoding).strip()
    rows.append({"file_name": f"S{session}", "file_content": text})

if not rows:
    print(f"No *_sentence.txt files found under {indir}")

with outfile.open("w", encoding=encoding, newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["file_name", "file_content"])
    writer.writeheader()
    writer.writerows(rows)

print(f"Wrote {len(rows)} rows to {outfile}")
```

```python
#!/usr/bin/env python3
"""
Split a list of conversation IDs into N evenly sized CSV files.

Input:
  - A CSV with a column containing the IDs (default column name: file_name).
    Example row: S001
Output:
  - group_1.csv, ..., group_N.csv with header: conversation_id

Usage:
  python split_into_groups.py --in data/processed/CANDOR/conversations.csv \
                              --column file_name \
                              --groups 8 \
                              --outdir data/processed/CANDOR

Options:
  --shuffle       Randomize order before splitting (off by default).
  --seed 42       Seed used when shuffling.
"""
import argparse
import csv
import math
import random
from pathlib import Path


def split_even(ids, n_groups: int):
    """
    Split ids into n_groups, distributing remainder one-by-one to the first groups.
    """
    total = len(ids)
    base = total // n_groups
    rem = total % n_groups
    groups = []
    start = 0
    for i in range(n_groups):
        size = base + (1 if i < rem else 0)
        groups.append(ids[start:start+size])
        start += size
    return groups

def write_groups(groups, outdir: Path, encoding="utf-8"):
    outdir.mkdir(parents=True, exist_ok=True)
    paths = []
    for i, g in enumerate(groups, start=1):
        p = outdir / f"group_{i}.csv"
        with p.open("w", encoding=encoding, newline="") as f:
            w = csv.writer(f)
            w.writerow(["conversation_id"])
            for x in g:
                w.writerow([x])
        paths.append(p)
    return paths


ids = list(dataset["file_name"])
groups = 4
seed=42
outdir = "./"
shuffle
if not ids:
    raise SystemExit("No IDs found in the input file.")

if shuffle:
    random.Random(seed).shuffle(ids)

groups = split_even(ids, groups)
paths = write_groups(groups, Path(outdir), encoding=encoding)
print(f"Wrote {len(paths)} files:")
for p in paths:
    print(" -", p)
```
